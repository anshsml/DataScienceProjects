```python
# Import dependencies
import sqlite3

# Connect to the SQLite database
conn = sqlite3.connect("data/petsdb")  
```

```python
# Check if connection is successful
def is_opened(conn):
    try:
        conn.execute("SELECT * FROM persons LIMIT 1")
        return True
    except sqlite3.ProgrammingError as e:
        print("Connection closed {}".format(e))
        return False
print(is_opened(conn))
```

```python
# Close the connection
conn.close()
```

```python
# Reconnect to the database
conn = sqlite3.connect("data/petsdb")
# Create a cursor object to interact with the database
c = conn.cursor()
```

```python
# Execute a query to count the number of people in each age group
c.execute("SELECT count(*), age FROM persons GROUP BY age")
# Fetch and print the results
results = c.fetchall()
for row in results:
    print("Age: {}, Count: {}".format(row[1], row[0]))
```

```python
# Get age group has the maximum number of people
c.execute("SELECT age, count(*) FROM persons GROUP BY age ORDER BY count(*) DESC LIMIT 1")
# Fetch and print the result
max_age_group = c.fetchone()
print("Age group with maximum people: {}, Count: {}".format(max_age_group[0], max_age_group[1]))
```

```python
# Get Number of people that do not have a last name
c.execute("SELECT count(*) FROM persons WHERE last_name IS NULL")
# Fetch and print the result
no_last_name_count = c.fetchone()[0]
print("Number of people without a last name: {}".format(no_last_name_count))
```

```python
# number of people have more than one pet
c.execute("""
    SELECT count(*)
    FROM persons
    WHERE id IN (
        SELECT owner_id
        FROM pets
        GROUP BY owner_id
        HAVING count(*) > 1
    )
""")
# Fetch and print the result
more_than_one_pet_count = c.fetchone()[0]
print("Number of people with more than one pet: {}".format(more_than_one_pet_count))
```

```python
#How many pets have received treatment
c.execute("SELECT count(*) FROM pets WHERE treatment_done IS NOT NULL")
# Fetch and print the result
treatment_count = c.fetchone()[0]
print("Number of pets that have received treatment: {}".format(treatment_count))
```

```python
# How many pets have received treatment, and the type of pet is known?
c.execute("""
    SELECT count(*)
    FROM pets
    WHERE treatment_done IS NOT NULL AND pet_type IS NOT NULL
""")
# Fetch and print the result
treatment_with_type_count = c.fetchone()[0]
print("Number of pets that have received treatment and type is known: {}".format(treatment_with_type_count))
```

```python
# How many pets are from the city called east port
c.execute("""
    SELECT count(*)
    FROM pets
    WHERE owner_id IN (
        SELECT id
        FROM persons
        WHERE city = 'east port'
    )
""")
# Fetch and print the result
east_port_pet_count = c.fetchone()[0]
print("Number of pets from the city called East Port: {}".format(east_port_pet_count))
```

```python

# How many pets are from the city called east port, and who received treatment?
c.execute("""
    SELECT count(*)
    FROM pets
    WHERE owner_id IN (
        SELECT id
        FROM persons
        WHERE city = 'east port'
    ) AND treatment_done IS NOT NULL
""")
# Fetch and print the result
east_port_treatment_count = c.fetchone()[0] 
print("Number of pets from East Port that received treatment: {}".format(east_port_treatment_count))
```

```python
# close the cursor and connection
c.close()
conn.close()
```
